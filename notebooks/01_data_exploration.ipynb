```python
import pandas as pd

# Load the VOLARE futures data
data = pd.read_csv("../data/raw/realized_variance_futures.csv")

# Keep ES only
es = data[data["symbol"] == "ES"].copy()

# Keep date and 5-minute realized variance
es["date"] = pd.to_datetime(es["date"])
es = es[["date", "rv5"]].dropna()

# Sort chronologically
es = es.sort_values("date").reset_index(drop=True)

print("Number of observations:", len(es))
print("Date range:", es["date"].min(), "to", es["date"].max())
print(es.head())
print(es.tail())
```
